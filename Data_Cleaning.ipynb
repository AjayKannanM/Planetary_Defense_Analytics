```python
id_cols = ['id', 'spkid', 'full_name', 'pdes', 'name', 'prefix']
df_clean = df_clean.drop(
    columns=[col for col in id_cols if col in df_clean.columns]
)
```

```python
df_clean.isnull().sum()
```

```python
if 'neo' in df_clean.columns:
  df_clean['neo'] = df_clean['neo'].map({'Y': 1, 'N': 0}).fillna(0)

if 'pha' in df_clean.columns:
  df_clean['pha'] = df_clean['pha'].map({'Y': 1, 'N': 0})
```

```python
if 'class' in df_clean.columns:
  df_clean = pd.get_dummies(df_clean, columns=['class'], drop_first=True)


df_clean = df_clean.select_dtypes(include=[np.number])
```

```python
numerical_cols = [
    col
    for col in df_clean.select_dtypes(include=[np.number]).columns
    if col not in ['pha', 'neo']
]

for col in numerical_cols:
  Q1 = df_clean[col].quantile(0.25)  # 25th percentile
  Q3 = df_clean[col].quantile(0.75)  # 75th percentile
  IQR = Q3 - Q1  # Interquartile range

  
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR

  df_clean[col] = np.where(
      df_clean[col] < lower_bound, lower_bound, df_clean[col]
  )
  df_clean[col] = np.where(
      df_clean[col] > upper_bound, upper_bound, df_clean[col]
  )

print('Outliers capped successfully via Winsorizing!')
```

```python
feature_cols_to_fill = [
    col
    for col in df_clean.select_dtypes(include=[np.number]).columns
    if col not in ['diameter', 'pha']
]

df_clean[feature_cols_to_fill] = df_clean[feature_cols_to_fill].fillna(
    df_clean[feature_cols_to_fill].median()
)

print(
    'Features filled with medians! Target nulls (like diameter) are preserved'
    ' for modeling.'
)
```
